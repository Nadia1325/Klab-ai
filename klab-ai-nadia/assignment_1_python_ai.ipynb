```python
# Normalise()

def normalise(values):
    """
    Scale values to the range [0, 1].
    A constant list is handled safely.
    """
    minimum = min(values)
    maximum = max(values)

    if maximum == minimum:
        return [0 for _ in values]

    return [(x - minimum) / (maximum - minimum) for x in values]
```

```python
values = [10,20,30,40,50]
normalised_values = normalise(values)

print("Original:", values)
print("Normalised:", normalised_values)
```

```python
constant_values = [5, 5, 5, 5]

print("Constant list", constant_values)
print("Normalised constant list:", normalise(constant_values))
```

```python
# summarise_score
def summarise_scores(scores):
    """
    return summary statistics for a list of scores.
    """
    return {
    "min": min(scores),
    "max": max(scores),
    "mean": sum(scores) / len(scores),
    "count": len(scores)  
    }
    
```

```python
scores = [65, 70, 80, 85, 90]
summary = summarise_scores(scores)
print("Summary:", summary)
```

```python
print("Minimum:", summary["min"])
print("Maximum:", summary["max"])
print("Mean:", summary["mean"])
print("Count:", summary["count"])
```

```python
#safe_divide
def safe_divide(a, b):
    """
    Divide a by b safely.
    Returns None if b is zero.
    """
    if b == 0:
        return None

    return a / b
```

```python
print("10 / 2 =", safe_divide(10, 2))
```

```python
print("10 / 0 =", safe_divide(10, 0))
```
